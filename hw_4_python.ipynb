```python
import re
```

```python
def check_car_id(car_id: str) -> str:
    res = re.match(r"^((?:[АВЕКМНОРСТУХ]|[ABEKMHOPCTX])\d{3}(?:[АВЕКМНОРСТУХ]|[ABEKMHOPCTX]){2})(\d{2,3})$", car_id)
    if res:
        print(f'Номер {res.group(1)} валиден. Регион: {res.group(2)}')
    else:
        print('Номер не валиден')
```

```python
check_car_id('А222ВС96')
```

```python
check_car_id('А123ВЕ777')
```

```python
check_car_id('АБ22ВВ193')
```
